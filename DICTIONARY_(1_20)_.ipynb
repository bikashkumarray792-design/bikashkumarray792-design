```python
#13.SUM OF ALL VALUES IN DICTIONARY
studm={"Math":80,"Python":90,"DBMS":75,"Java":85}
sum=0
for i in studm:
     sum=sum+studm[i]
print("SUM OF MARKS :",sum)
```

```python
#16.COUNT OF FREQUENCY OF EACH CHARACTER IN STRING USING DICTIONARY

s="python"
k="agreement"
l=list(s)
l1=list(k)
d1={}
d2={}
for i in l:
    d1[i]=(l.count(i))
print(d1)
for i in l1:
    d2[i]=(l1.count(i))
print(d2)
```

```python
#17.COUNT OF FREQUENCY OF EACH ELEMENT IN LIST USING DICTIONARY
l=['apple','ball','apple','cat']
print(l)
d={}
for i in l:
    d[i]=l.count(i)
print(d)
```

```python
#18.FIND DUPLICATE ELEMENTS IN LIST USING DICTIONARY
l=['apple','ball','apple','cat']
print(l)
d={}
for i in l:
    q=l.count(i)
    if q>1:
      d[i]=q
print("Duplicate Elements:",d)
```

```python
#19.MERGING TWO DICTIONARIES
stud={"Name":'Bikash',"Age":21,"Course":"Python"}
studm={"Math":80,"Python":90,"DBMS":75,"Java":85}
stud.update(studm)
print(stud)
```

```python
#20.SORTING OF DICTIONARY USING ITS VALUES

studm={"Math":80,"Python":90,"DBMS":75,"Java":85}
d={}
print(studm)
while studm:
    for i in studm:
        s=i
        svalue=studm[i]
        break

    for i in studm:
        if studm[i]<svalue:
            svalue=studm[i]
            s=i

    d[s]=svalue
    del studm[s]

print("SORTED DICTIONARY",d)
```
