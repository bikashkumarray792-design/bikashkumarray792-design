```python
#1. SIMPLE SHOPPING CART

print("**** GROCERY SHOP ****")
ip = [(1,'rice',50),(2,'Water',20),(3,'ToyCar',100),(4,'Chocolate',50),(5,'pen',20)]

print("\nSl. Items \t Prices")

k = []
i = []
p = []

for s, items, prices in ip:
    k.append(s)
    i.append(items.lower())
    p.append(prices)
    print(s, ".", items.upper(), "\t", prices)

cart = []
total = 0
ch=0
c = input("\n Enter Items (use a comma after enter one item): ").lower().split(",")

for itm in c:
    itm=itm.strip()

    if itm in i:
        idx=i.index(itm)
        pr=p[idx]

        cart.append((itm,pr))
        total=total+pr
        ch+=1

    else:
        print(itm, "not found")
if ch!=0:
   print("\n  \t  CART\n")
   print("ITEMS".ljust(15), "PRICE")
   for i, p in cart:
      print(i.upper().ljust(15), p)
   print("\nTOTAL = ₹", total)
   print("\nTHANK YOU, VISIT AGAIN !")
else:
   print("")
```

```python
#4. MINI ATM TRANSACTION SYSTEM

print("\t\tATM TRANSACTION SYSTEM\n")

trans=[]
bal=0

while True:
    print("\n1. DEPOSIT")
    print("2. WITHDRAW")
    print("3. TRANSACTION HISTORY")
    print("4. CHECK BALANCE")
    print("5. EXIT")

    ch=int(input("\nEnter your choice: "))

    if ch==1:
        amount=int(input("Enter deposit amount: "))

        if amount > 0:
            bal= bal + amount
            trans.append(("Deposit", amount))
            print("Amount deposited successfully!")
        else:
            print("Invalid amount!")

    elif ch==2:
        amount = int(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Invalid amount!")
        elif amount > bal:
            print("Insufficient balance!")
        else:
            bal= bal - amount
            trans.append(("Withdraw", amount))
            print("Amount withdrawn successfully!")

    elif ch == 3:
        print("\nTRANSACTION HISTORY")
        print("TYPE".ljust(15), "AMOUNT")

        for typ, amount in trans:
            print(typ.ljust(15), amount)

    elif ch == 4:
        print("\nCurrent Balance: ₹", bal)

    elif ch == 5:
        print("\nThank you for using ATM!")
        break

    else:
        print("Invalid choice!")
```

```python
#5. FOOD ORDERING SYSTEM

print("*** FOOD ORDERING SYSTEM ***\n")

menu=[("Burger",120),("Pizza",250),("Momos",100),("Cold Drink",50)]

ord= []
total = 0

print("\t MENU \t\n")
print("ITEM".ljust(15), "PRICE")

for itm,pr in menu:
    print(itm.ljust(15),pr)

while True:
    ch=input("\nEnter item name (or 'done' to finish): ")

    if ch.lower()=="done":
        break

    found=False

    for itm,pr in menu:
        if itm.lower()==ch.lower():
            ord.append((itm, pr))
            total=total+pr
            print(itm, "added to order")
            found=True
            break

    if found==False:
        print("Item not found!")

print("\n     YOUR ORDER\n")
print("ITEM".ljust(15), "PRICE")

for itm,pr in ord:
    print(itm.ljust(15), pr)

print("\nTOTAL BILL = ₹", total)
print("Thank you for ordering!")
```

```python
#6. LIBRARY BOOK MANAGEMENT

print("**** LIBRARY BOOK MANAGEMENT ****")

books = [
    ["Python", "Available"],
    ["AI", "Issued"],
    ["Machine Learning", "Available"]
]

while True:

    print("\n1.DISPLAY ALL BOOKS")
    print("2.SEARCH BOOK")
    print("3.ISSUE BOOK")
    print("4.RETURN BOOK")
    print("5.EXIT")

    ch=int(input("\nEnter your choice: "))
    if ch==1:
        print("\nBOOKS")
        print("\nBOOKS".ljust(20),"STATUS")
        for book,status in books:
            print(book.ljust(20),status)


    elif ch==2:
        name=input("Enter book name: ")
        found=False

        for book,status in books:
            if book.lower()==name.lower():
                print("Book:", book)
                print("Status:", status)
                found=True

        if found==False:
            print("Book not found!")


    elif ch==3:
        name=input("Enter book name to issue: ")
        found=False

        for bookn in books:
            if bookn[0].lower()==name.lower():
                found=True

                if bookn[1]=="Available":
                    bookn[1]="Issued"
                    print("Book issued successfully!")
                else:
                    print("Book is already issued!")

        if found==False:
            print("Book not found!")

    # RETURN
    elif ch==4:
        name=input("Enter book name to return: ")
        found=False

        for bookr in books:
            if bookr[0].lower()==name.lower():
                found=True

                if bookr[1]=="Issued":
                    bookr[1]="Available"
                    print("Book returned successfully!")
                else:
                    print("Book is already available!")

        if found==False:
            print("Book not found!")

    # EXIT
    elif ch==5:
        print("Thank you for using Library!")
        break

    else:
        print("Invalid choice!")
```

```python
#7. STUDENTS RESULT SYSTEM

print("**** STUDENTS RESULT SYSTEM ****")

students=[("Amit", 85, 78, 90),("Ram",89,75,91),("Smita", 80, 75, 90),("Abhijit", 85, 88, 92),("Rakesh", 70, 65, 78),("Rekha",30,29,79)]
high=0
top=""

for name,m1,m2,m3 in students:
    total=m1+m2+m3
    percentage=total/3
    if percentage>=80:
        grade="A"
    elif percentage>=60:
        grade="B"
    elif percentage>=40:
        grade="C"
    else:
        grade="F"

    if m1>=30 and m2>=30 and m3>=30:
        result="PASS"
    else:
        result="FAIL"

    print("NAME:",name)
    print("MARKS:",m1,m2,m3)
    print("TOTAL:",total)
    print("PERCENTAGE:",f"{percentage:.2f}","%")
    print("GRADE:",grade)
    print("RESULT:",result)
    print("-------------------------")

    if total>high:
        high=total
        top=name

print("TOPPER:",top)
print("HIGHEST TOTAL:", high)
```

```python
#8. MOVIE RATING SYSTEM
print("**** MOVIE RATING SYSTEM ****\n")


movies=[("Inception", 9),("3 Idiots", 9.2),("Dangal", 8.3)]

while True:
    print("\n1. ADD MOVIE")
    print("2. DISPLAY MOVIES")
    print("3. SEARCH MOVIE")
    print("4. HIGHEST RATED MOVIE")
    print("5. EXIT")

    ch=int(input("\nEnter your choice: "))

    if ch==1:
        name=input("Enter movie name: ")
        rating=float(input("Enter rating: "))

        movies.append((name, rating))
        print("Movie added successfully!")

    elif ch==2:
        print("\nMOVIES")
        print("MOVIE".ljust(20), "RATING")

        for name,rating in movies:
            print(name.ljust(20), rating)

    elif ch==3:
        search=input("Enter movie name: ")
        found=False

        for name,rating in movies:
            if name.lower()==search.lower():
                print("Movie:", name)
                print("Rating:", rating)
                found=True

        if found==False:
            print("Movie not found!")

    elif ch==4:
        high=0
        movie=""

        for name,rating in movies:
            if rating>high:
                high=rating
                movie=name

        print("Highest Rated Movie:",movie)
        print("Rating:",high)

    elif ch==5:
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```

```python
#9. CRICKET SCORE TRACKER
print("**** CRICKET SCORE TRACKER ****\n")


players=[("Virat", 85),("Rohit", 72),("Rahul", 45)]

while True:
    print("\n1. DISPLAY SCORES")
    print("2. TEAM TOTAL")
    print("3. HIGHEST SCORER")
    print("4. AVERAGE SCORE")
    print("5. EXIT")

    ch=int(input("\nEnter your choice: "))

    if ch==1:
        print("\nPLAYER".ljust(15), "SCORE")

        for name, score in players:
            print(name.ljust(15), score)

    elif ch==2:
        total=0

        for name,score in players:
            total=total+score

        print("TEAM TOTAL:", total)

    elif ch==3:
        highest=0
        player=""

        for name,score in players:
            if score>highest:
                highest=score
                player=name

        print("HIGHEST SCORER:",player)
        print("SCORE:",highest)

    elif ch==4:
        total=0

        for name,score in players:
            total=total+score

        average=total/len(players)

        print("AVERAGE SCORE:",f"{average:.2f}")

    elif ch==5:
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```

```python
#10. SIMPLE BILLING SYSTEM
print("**** SIMPLE BILLING SYSTEM ****\n")


items=[("Pen", 10, 5),("Notebook", 50, 2)]

while True:
    print("\n1. DISPLAY BILL")
    print("2. ADD ITEM")
    print("3. EXIT")

    ch=int(input("\nEnter your choice: "))

    if ch==1:
        total=0

        print("\nITEM".ljust(15), "PRICE".ljust(10), "QTY".ljust(10), "AMOUNT")

        for item, price, quantity in items:
            amount=price*quantity
            total=total+amount

            print(item.ljust(15), str(price).ljust(10),
                  str(quantity).ljust(10),amount)

        print("\nTOTAL BILL = ₹",total)

    elif ch==2:
        name=input("Enter item name: ")
        price=int(input("Enter price: "))
        quantity=int(input("Enter quantity: "))

        items.append((name, price, quantity))

        print("Item added successfully!")

    elif ch==3:
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
