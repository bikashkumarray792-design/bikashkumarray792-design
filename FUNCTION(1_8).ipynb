```python
#1.CREATE A FUNCTION TO FIND SQUARE OF A NUMBER
def sqr(n):
    return n**2
s=int(input("ENTER A NUMBER:"))
print("SQUARE OF ",s,"IS:",sqr(s))
```

```python
#2. NUMBER IS EVEN OR ODD
def evenodd(n):
  if n%2==0:
    print(n," IS EVEN")
  else:
    print(n," IS ODD")
n=int(input("ENTER A NUMBER:"))
evenodd(n)
```

```python
#3. LARGEST NUMBER IN A LIST
def largest(n):
  lar=n[0]
  for i in n:
    if i>lar:
      lar=i
  return lar

l=[1,2,3,4]
print("LARGEST VALUE IN ",l,"IS :",largest(l))
```

```python
#4. CALCULATE AVERAGE OF NUMBERS IN A LIST
def avgg(n):
  sum=0
  for i in n:
    sum+=i
  return sum/len(n)

l={1,2,3,4,5}
print("AVERAGE OF NUMBERS IN LIST ",l,"IS :",avgg(l))
```

```python
#5. COUNT VOWELS IN A STRING
def countvowel(t):
  cnt=0
  v=['A','a','E','e','I','i','O','o','U','u']
  for i in t:
    if i in v:
      cnt+=1
  return cnt
s="apple"
L=list(s)
print("NUMBER OF VOWELS IN",s.upper(),"IS : ",countvowel(L))
```

```python
#6. AREA OF CIRCLE
def area(n):
  return 3.14*n**2
r=int(input("ENTER RADIUS OF CIRCLE:"))
print("AREA OF CIRCLE WITH RADIUS ",r,"IS :",area(r))
```
