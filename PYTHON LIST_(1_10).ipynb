```python
#1. LARGEST NUMBER
numbers = [100,11,-1,-10,55,67]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest element:", largest)
```

```python
#2.SMALLEST NUMBER
numbers = [12, 45, 2, 89, 33, 7]

smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num

print("Smallest element :", smallest)
```

```python
#3.SECOND LARGEST

n=[10,63,99,12,-65,17]
n.sort(reverse=True)
print("Second largest:",n[1])
```

```python
#4.SECOND SMALLEST

n=[10,63,99,12,-65,17]
n.sort()
print("Second Smallest:",n[1])
```

```python
#5.REVERSE LIST
n=[10,63,99,12,-65,17]
rev=[]
for i in range(len(n)-1,-1,-1):
     rev.append(n[i])

print(rev)
```

```python
#6. LIST WITHOUT DUPLICATE ELEMENTS

n=[15,52,342,17,52,342,4,15]
nonrep=[]
for i in n:
    if i not in nonrep:
        nonrep.append(i)

print("List without duplicates:",nonrep)
```

```python
#7.FIND DUPLICATE ELEMENTS

n=[15,52,342,17,52,342,4,15]
u=[]
d=[]

for i in n:
    if i in u:
        if i not in d:
            d.append(i)
    else:
        u.append(i)

print(d)
```

```python
#8.SUM OF ALL ELEMENTS

n = [15,52,342,17,52,342,4,15]
s=0
for i in n:
    s+=i
print("sum of all elements is:",s)
```

```python
#9.AVERAGE OF ALL ELEMENTS

n = [15,52,342,17,52,342,4,15]
s=0
for i in n:
    s+=i
avg=s/len(n)
print("Average of all elements is:",avg)
```

```python
#10.NO OF EVEN AND ODD

n=[10,16,62,77,-12,49,37]
evenn=0
oddn=0

for num in n:
    if num%2==0:
        evenn+= 1
    else:
        oddn+= 1
print("No. Of Evennumbers:", evenn)
print("No. Of Oddnumbers:",oddn)
```
