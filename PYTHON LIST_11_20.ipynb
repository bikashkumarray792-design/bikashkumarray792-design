```python
#11. SEARCH FOR ELEMENTS

n=[17,3,20,5,12,-12,100]
s=int(input("Enter number to search: "))
if s in n:
    print("Element found")
else:
    print("Element not found")
```

```python
#12. OCCURANCE OF NUMBER

n=[10,19,25,27,35,25,10]
r=int(input(print("Enter number from",n," :")))
c=0
for i in n:
    if i==r:
        c+=1
print("Occurrence of number",r," =",c)
```

```python
#13. COMMON ELEMENTS BETWEEN TWO LISTS

n1=[1,20,15,30,100]
n2=[2,20,30,19,17]
n3=[]

for i in n1:
    if i in n2:
       n3.append(i)
print("Common Elements in",n1,"and",n2,"are :\n",n3)
```

```python
#14.MERGE TWO LISTS

n1=[1,20,15,30]
n2=[2,21,45]
n3=[]
n3=n1+n2

print("After Merging of",n1,"and",n2,"are:\n",n3)
```

```python
#15.SORT LIST WITHOUT SORT()

n=[50,-17,23,200,19]
for i in range(len(n)):
    for j in range(i + 1, len(n)):
        if n[i] > n[j]:
            temp = n[i]
            n[i]=n[j]
            n[j]=temp

print(n)
```

```python
#16. LIST IS PALINDROME OR NOT

n1=[17,3,20,5]
n2=n1.reverse()

n3=[17,19,10,6,6,10,19,17]
n4=n3.reverse()

if n1==n2:
   print(n1," is a Palindrome\n")
else:
   print(n1,"is not a Palindrome\n")

if n3==n4:
   print(n3,"is a Palindrome")
else:
   print(n3,"is not a Palindrome")
```

```python
#17. MISSING NUMBER FROM 1 - N

n1=[1,2,3,4,6]
n2=6
t=0

for i in range(1,n2+1):
        t=t+i
for i in n1:
        t=t-i
print("Missing number from",n1,"is :",t)
```

```python
#18.MOVE ALL ZERO TO THE END
n=[0,1,2,0,3,0,4]
r=[]

for i in n:
    if i!=0:
      r.append(i)
for i in n:
    if i==0:
      r.append(i)

print("After Moving all zero to end: ",r)
```

```python
#19. FIRST REPEATED ELEMENT

n=[1, 23, 45, 23, 45, 60]
for i in range(len(n)):
    for j in range(i+1,len(n)):
        if n[i]==n[j]:
            print("First repeated element in",n,"= ",n[i])
            break
    else:
        continue
    break
```

```python
#20.FIRST NON-REPEATED ELEMENT
n=[1,23,45,23,45]
for i in n:
    c=0
    for j in n:
        if i==j:
            c=c+1

    if c==1:
        print("First non-repeated element in",n,"= ", i)
        break
```
