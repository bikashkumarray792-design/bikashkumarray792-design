```python
#11.LIST TO TUPLE

li=[1,23,'c','java',456]
tup=tuple(li)
print("Tuple is: ",tup)
```

```python
#12. TUPLE TO LIST
tup=(1,23,'c','java',456)
lis=[]
for i in tup:
    lis.append(i)
print("List is: ",lis)
```

```python
#13.REVERSE A TUPLE
t=(1,23,15,20,456)
l=[]
for i in t:
   l.append(i)
k=[]
for i in range(len(l)-1,-1,-1):
      k.append(l[i])
tup=tuple(k)
print("REVERSE OF TUPLE",t," is: ",tup)
```

```python
#14.CONCATE TWO TUPLES
t1=(1,23,45,'c','java','python')
t2=(67,89,'php','sql')
print("CONACTINATION OF ",t1,"AND ",t2,"IS :",t1+t2)
```

```python
#15.COMMON ELEMENTS BETWEEN TWO TUPLES
t1=(1,23,'c','java')
t2=('c','java',23,19)
l=[]
c=0

for i in t1:
    if i in t2:
       l.append(i)
       c+=1

if c==0:
   print("NO COMMON ELEMENTS BETWEEN ",t1," AND ",t2)
else:
   print("COMMON ELEMENTS BETWEEN",t1,"AND",t2,"IS :",l)
```

```python
#16.REMOVE DUPLICATE ELEMENTS
t=(1,23,'c','java',23,'c',45)
l=[]
for i in t:
   l.append(i)
u=[]
d=[]

for i in l:
   if i not in u:
      u.append(i)

   else:
      d.append(i)
      c+=1
t1=tuple(d)
t2=tuple(u)
if c==0:
   print("NO DUPLICATE ELEMENTS IN",t)
else:
   print("AFTER REMOVING DUPLICATE ELEMENTS IN ",t," ARE :",t2,"\n AND DUPLICATE ELEMENTS ARE :",t1)
```

```python
#17.MAXIMUM AND MINIMUM WITHOUT MAX() AND MINIMUM()

t=(1,23,64,-17)
L=sorted(t)
print("\n MAXIMUM ELEMENT IN TUPLE",t," IS:",L[len(L)-1])
print("\n MINIMUM ELEMENT IN TUPLE",t," IS:",L[0])
```

```python
#18.COUNTING OF EVEN AND ODD NUMBERS
t=(1,23,20,17,5,3,12,64)
ec,oc=0,0
et,ot=[],[]

for i in t:
    if i%2==0:
      et.append(i)
      ec+=1

    else:
      ot.append(i)
      oc+=1
print("THERE ARE ",ec," EVEN ELEMENTS ARE PRESENT IN TUPLE ",t," THEY ARE :",tuple(et))
print("THERE ARE ",oc," ODD ELEMENTS ARE PRESENT IN TUPLE ",t," THEY ARE :",tuple(ot))
```

```python
#19.CHECK FOR TUPLE IS PALINDROME OR NOT
t=(1,23,64,17,3,5,19,7,8)
ts=(1,21,21,1)
l=[]
l1=[]
for i in range(len(t)-1,-1,-1):
     l.append(t[i])
for i in range(len(ts)-1,-1,-1):
     l1.append(ts[i])
t1=tuple(l)
t2=tuple(l1)
if t==t1:
   print("\n\nTUPLE ",t," IS PALINDROME.")
else:
   print("\n\nTUPLE ",t," IS NOT PALINDROME.")
if ts==t2:
   print("\n\nTUPLE ",ts," IS PALINDROME.")
else:
   print("\n\nTUPLE ",ts," IS NOT PALINDROME.")
```
