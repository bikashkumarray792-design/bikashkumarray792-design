```python
#1.CREATE AND DISPLAY TUPLE
T1=()
T2=("c","java","python")
print("\nEmpty Tuple is:",T1)
print("\ntupple is:",T2)
```

```python
#2.LENGTH OF TUPLE
t=("c","python",1,2)
print("\nLength Of Tuple is : ",len(t))
```

```python
#3.FIRST AND LAST ELEMENT OF TUPLE
t=("c","python","java",1,2)
print("\nFirst element of the tuple ",t," is: ",t[0])
print("\n Last element of the tuple ",t," is: ",t[len(t)-1] )
```

```python
#4.LARGEST ELEMENT
t=(1,23,45,64,-1)
l=sorted(t)
print("\nLargest Element in",t," is:",l[len(l)-1])
```

```python
#5.SMALLEST ELEMENT
t=(1,23,45,67,-1,-17)
l=sorted(t)
print("\nSmallest Element in",t," is:",l[0])
```

```python
#6.SUM OF ALL ELEMENT
t=(1,23,45,67,-1,-17)
l=sum(t)
print("\nsum of all element in",t," is:",l)
```

```python
#7.AVERAGE OF ALL ELEMENT
t=(1,23,45,67,-1,-17)
a=sum(t)/len(t)
print("\n Average sum of all element in",t," is:",a)
```

```python
#8.COUNT OCCURANCE OF AN ELEMENT
t=(1,10,9,8,10,12)
print("Tuple is ",t )
r=int(input("\n ENTER THE ELEMENT FROM TUPLE IS: "))
c=0
for i in t:
    if i == r:
      c+=1

print("\n Occurance of ",r," is:",c)
```

```python
#9.INDEX OF ELEMENT
t=(1,10,9,8,45,-1)
print("Tuple is ",t)
i=int(input("Enter the element from tuple whose index you want to find: "))
print("Index of ",i," in the tuple ",t,"is :",t.index(i))
```

```python
#10.ELEMENT EXISTANCE
t=(1,10,23,45,-1,17)
e=int(input("Enter a number whose existance you want to find :" ))
c=0
for i in t:
    if i==e:
         c+=1

if c==1:
   print("Element ",e," is exist. In the tuple ",t)
else:
   print("Element ",e," is not exist.In the tuple",t)
```
